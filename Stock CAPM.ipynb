```python
company_name = input("Please enter the name of the stock: ")

stock_beta = float(input("Enter the stock beta: "))

risk_free_rate = float(input("Enter the risk-free rate of return in %: "))

market_rate = float(input("Enter the expected broad market rate of return in % (Ex: S&P500): "))

expected_return = (risk_free_rate + stock_beta*(market_rate-risk_free_rate))

print("Using the CAPM formula, the expected return on {} stock = {}%".format(company_name, expected_return))
```

```python
dividend= float(input("Enter this year's dividend: "))

growth_rate= float(input("Enter the company's expected dividend growth: "))

req_rate= float(input("Enter the company's cost of equity: "))

stock_price= float(input("Enter the current stock price: "))

expected_dividend= dividend*(1+(growth_rate/100))
print(expected_dividend)

dividend_discount_model=(expected_dividend/((req_rate-growth_rate)/100))
print("$",dividend_discount_model)

if dividend_discount_model>stock_price:
    print("Stock is undervalued, invest in it!")

elif dividend_discount_model<stock_price:
    print("Stock is overvalued, don't invest in it!")
          
else:
    print("Stock is fairly valued")
```
